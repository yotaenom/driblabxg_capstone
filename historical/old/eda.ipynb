```python
import pandas as pd
import pathlib
import json

# Define the directory
jsonl_dir = pathlib.Path("data/shot_pack/jsonls")

# Read and concatenate all JSONL files
dfs = []
for file in jsonl_dir.glob("*.jsonl"):
    with open(file, "r") as f:
        lines = [json.loads(line) for line in f]
        dfs.append(pd.DataFrame(lines))

# Combine into one DataFrame
df_all = pd.concat(dfs, ignore_index=True)

# Write to Parquet
df_all.to_parquet("all_tracking_data.parquet", index=False)
```
